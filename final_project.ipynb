#**Section 3: Data Cleaning & Exploratory Data Analysis (EDA)**

#A. Data cleaning

**3.1. Handling Missing Value**

```python
# عرض القيم الناقصة قبل المعالجة
data.isnull().sum()
```

```python
#  إنشاء دالة لتحديد القريد من الفاينل سكور
def assign_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'

        #تعويض القيم الناقصة في القريد
data['grade'] = data['grade'].fillna(data['final_score'].apply(assign_grade))
```

Compared to dropping rows, the chosen methods preserve all data without losing information.
The custom method provides higher accuracy for the `grade` column, while mode imputation is suitable for categorical features.

```python
#  تعويض القيم الناقصة في الأعمدة التصنيفية باستخدام الـ Mode
data['parent_education_level'].fillna(data['parent_education_level'].mode()[0], inplace=True)
data['study_environment'].fillna(data['study_environment'].mode()[0], inplace=True)
```

```python
# عرض القيم الناقصة بعد المعالجة
data.isnull().sum()
```

```python
# عدد الصفوف بعد
print(data.shape)
```

Two different methods were applied to handle missing values.

For the `grade` column, a custom method was used by assigning grades based on the `final_score`. This approach ensures logical consistency between the two variables.

For categorical variables such as `parent_education_level` and `study_environment`, mode imputation was applied by filling missing values with the most frequent category.

Although other methods such as dropping rows or using mean were considered, they were not used because the number of missing values is small and removing data could lead to unnecessary information loss.

**3.7 Data Aggregation**


Aggregation techniques were applied to summarize the data and analyze relationships between variables.

These methods help identify patterns and differences between groups in the dataset.

```python
# حساب متوسط الدرجة النهائية لكل بيئة دراسة
data.groupby('study_environment')['final_score'].mean()

# إنشاء جدول محوري لمقارنة متوسط الدرجة حسب القريد وبيئة الدراسة
pd.pivot_table(
    data,
    values='final_score',
    index='grade',
    columns='study_environment'
)
```


The aggregation results reveal clear relationships between student performance and behavioral factors.

From the analysis grouped by `final_score`, it is observed that higher-performing students tend to have higher daily study hours, better attendance percentages, higher homework completion rates, and higher motivation scores. This indicates a strong positive relationship between these factors and academic performance.

Students with lower final scores show noticeably lower averages in these same factors, suggesting that lack of consistency in studying, attendance, and motivation negatively impacts performance.

Additionally, the pivot table comparing grades across different study environments shows that students perform slightly better in certain environments, although the differences are relatively small. This suggests that while study environment has some influence, behavioral factors such as effort and consistency play a more significant role.

Overall, the results highlight that student habits—such as studying regularly, attending classes, completing homework, and maintaining motivation—are key drivers of academic success.

#**Section 5: Data Transformation**

**5.1. Data Normalization and Scaling (Numerical Features)**

__Importance of Normalization and Scaling__

* Many ML algorithms rely on distance

* Features with large values dominate over smaller ones. For example, if one feature has a value between 0 and 1000 and another between 0 and 1, the larger one will dominate

* Scaling improves model performance and convergence speed

### **Types of Normalization Techniques**



___Numerical Features___:

* __stars__  If stars is the Target variable --> **X** we do not scale it

* __useful__


### **1.Normalization (Min-Max Scaling)**

Min-max scaling, also known as rescaling, is a popular normalization technique that rescales the data to a common range, usually between 0 and 1. This is achieved by subtracting the minimum value and then dividing by the range of the data.


# $x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}$

Where:

- $( x $) = original value  
- $( x_{min} $) = minimum value in the feature  
- $( x_{max} $) = maximum value in the feature  
- $( x_{scaled})$ in [0,1] -> the new scaled value, which will lie between 0 and 1


### **2. Z-score Standardization**

Z-score normalization, also known as standardization, transforms data into a standard normal distribution with a mean of 0 and a standard deviation of 1. This technique is useful when the data follows a normal distribution.


# $ z = \frac{x - \mu}{\sigma}$

Where:

- $( x $) = original value  
- $( \mu $) = mean of the feature  
- $( \sigma $) = standard deviation  
- $( z $) = standardized value



---



Z-score scaling was selected over Min-Max normalization because it is more robust to outliers and standardizes the data for better model performance.

#6.3. Feature Extraction (PCA)



Principal Component Analysis (PCA) was applied to reduce the dimensionality of the dataset while preserving most of the important information.

PCA transforms the original features into a new set of uncorrelated components called principal components, which capture the maximum variance in the data.

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# اختيار الأعمدة الرقمية فقط
numeric_data = data_transformed.select_dtypes(include=['float64', 'int64'])

# عمل Scaling قبل PCA (ضروري)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_data)

# تطبيق PCA
pca = PCA()
pca_result = pca.fit_transform(scaled_data)

# عرض نسبة التباين لكل مكون
explained_variance = pca.explained_variance_ratio_

print("Explained Variance Ratio:")
print(explained_variance)

# حساب التباين التراكمي
import numpy as np
cumulative_variance = np.cumsum(explained_variance)

print("\nCumulative Variance:")
print(cumulative_variance)
```

### PCA Explanation and Analysis

PCA was applied to reduce the number of features and handle multicollinearity between highly correlated variables such as previous scores. It transforms the original features into a smaller set of principal components that capture the most important patterns in the data.

The explained variance ratio shows that the first principal component captures approximately 39.7% of the total variance, while the next few components contribute additional important information. Based on the cumulative variance, around 6–7 components are sufficient to preserve more than 85–90% of the original data information.

This demonstrates that PCA effectively reduces dimensionality while retaining most of the important information. As a result, it simplifies the dataset, removes redundancy between correlated features, and improves model efficiency and performance.



---



**Explained Variance (Elbow Plot)**

```python
import matplotlib.pyplot as plt
import numpy as np

# حساب التباين التراكمي
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)

plt.figure()
plt.plot(cumulative_variance, marker='o')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA - Explained Variance Curve')
plt.show()
```

### PCA Explained Variance Curve Analysis

The PCA explained variance curve shows how much of the total variance is captured as the number of principal components increases.

From the plot, the first principal component captures around 40% of the variance. As more components are added, the cumulative explained variance increases rapidly. The curve shows a clear elbow around 5 to 7 components.

By the time we reach:
- 3 components → approximately 80% of the variance
- 5 components → approximately 93%
- 6–7 components → more than 95%

After this point, adding more components results in only small improvements, indicating diminishing returns.

This suggests that most of the important information in the dataset can be retained using a reduced number of components (around 6–7), which significantly reduces dimensionality while preserving the majority of the data structure.

Overall, the PCA curve confirms that dimensionality reduction is effective and that only a few components are sufficient to represent the dataset efficiently.
